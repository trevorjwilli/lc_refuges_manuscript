```python
import pandas as pd
```

```python
def extract_data_from_log(file_path: str):
    """
    Function to extract the `per_barcode_raw_read_counts` data
    from the stacks process_radtags log
    """

    with open(file_path, 'r') as f:
        index = 0
        data = []
        for line in f:
            line = line.strip()

            if line == "BEGIN effective_coverages_per_sample":
                index += 1
            elif line == "END effective_coverages_per_sample":
                break
            elif index == 1:
                if '#' in line:
                    next
                else:
                    colnames = line.split('\t')
                    index += 1
            elif index > 1:
                data.append(line.split('\t'))

        out = pd.DataFrame(data, columns=colnames)
    return out
```

```python
df = extract_data_from_log('../denovo.all.pops/gstacks.log.distribs')
df['mean_cov_ns'] = df['mean_cov_ns'].astype(float)
```

```python
df_fil = df[df['mean_cov_ns'].astype(float) >= 5].reset_index(drop=True)
```

```python
import re

df_fil['population'] = df_fil['sample'].apply(lambda x: re.sub('[0-9]+_(.+)_LTC.+', '\\1', x))
```

```python
df_bishop = df_fil[df_fil['population'].isin(['Bishop', 'Red_Knolls'])].reset_index(drop=True)
df_gandy = df_fil[df_fil['population'].isin(['Gandy', 'Keg_Springs'])].reset_index(drop=True)
df_leland = df_fil[df_fil['population'].isin(['Leland_Harris', 'Lower_Rocky'])].reset_index(drop=True)
df_clear = df_fil[df_fil['population'].isin(['Clear_Lake', 'Willow_Pond'])].reset_index(drop=True)
df_mills = df_fil[df_fil['population'].isin(['Mills_Valley', 'Cluster', 'Copilot', 'Rosebud_Top_P'])].reset_index(drop=True)
df_mona = df_fil[df_fil['population'].isin(['Mona', 'Big_Springs', 'Jail_Pond', 'Tooele_Army'])].reset_index(drop=True)
```

```python
df_fil[['sample', 'population']].to_csv('../popmap_final.tsv', sep='\t', index=False, header=False)
df_bishop[['sample', 'population']].to_csv('../popmap_bishop.tsv', sep='\t', index=False, header=False)
df_gandy[['sample', 'population']].to_csv('../popmap_gandy.tsv', sep='\t', index=False, header=False)
df_leland[['sample', 'population']].to_csv('../popmap_leland_harris.tsv', sep='\t', index=False, header=False)
df_clear[['sample', 'population']].to_csv('../popmap_clear_lake.tsv', sep='\t', index=False, header=False)
df_mills[['sample', 'population']].to_csv('../popmap_mills_valley.tsv', sep='\t', index=False, header=False)
df_mona[['sample', 'population']].to_csv('../popmap_mona.tsv', sep='\t', index=False, header=False)
```
